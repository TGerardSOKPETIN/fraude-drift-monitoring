# Exploration : détection de data drift sur Poundbank

Ce notebook reprend le script `explorer_drift_detection.py`, découpé en cellules pour pouvoir exécuter et lire les résultats étape par étape (Exécuter avec Maj+Entrée, cellule par cellule).

Trois outils, l'un après l'autre : **Evidently AI**, **Alibi-Detect**, **NannyML**.

Pré-requis (dans ce même environnement) :
```
pip install pandas numpy scipy evidently alibi-detect nannyml
```

Place `reference.csv` et `analysis.csv` dans le même dossier que ce notebook avant de lancer la cellule suivante.

````
from pathlib import Path
import numpy as np
import pandas as pd

DATA_DIR = Path.cwd()  # adapte si tes CSV sont ailleurs

TIMESTAMP_COL = "timestamp"
TARGET_COL = "is_fraud"
PREDICTION_COL = "predicted_fraud"
PREDICTED_PROBA_COL = "predicted_fraud_proba"

NUMERICAL_FEATURES = ["time_since_login_min", "transaction_amount", "user_tenure_months"]
CATEGORICAL_FEATURES = ["transaction_type", "is_first_transaction"]
FEATURE_COLUMNS = NUMERICAL_FEATURES + CATEGORICAL_FEATURES

print("Configuration chargée.")
````

## 0. Chargement des données

````
reference = pd.read_csv(DATA_DIR / "reference.csv", parse_dates=[TIMESTAMP_COL])
analysis = pd.read_csv(DATA_DIR / "analysis.csv", parse_dates=[TIMESTAMP_COL])

print(f"reference.csv : {len(reference)} lignes")
print(f"analysis.csv  : {len(analysis)} lignes")
reference.head()
````

## 1. Evidently AI — Data Drift preset

Evidently choisit automatiquement KS, Chi-deux ou PSI selon le type de chaque colonne et compare `reference` à `analysis`.

````
from evidently import Report
from evidently.presets import DataDriftPreset

report = Report(metrics=[DataDriftPreset()])
evidently_result = report.run(
    reference_data=reference[FEATURE_COLUMNS],
    current_data=analysis[FEATURE_COLUMNS],
)

evidently_result.save_html("evidently_data_drift.html")
print("Rapport HTML généré : evidently_data_drift.html (ouvre-le dans un navigateur)")
````

````
# Aperçu direct dans le notebook (Jupyter sait afficher le rapport Evidently inline)
evidently_result
````

## 2. Alibi-Detect — TabularDrift (KS + Chi-deux, correction FDR)

Test KS sur chaque variable numérique, Chi-deux sur chaque variable catégorielle, avec correction pour tests multiples car on teste plusieurs colonnes en parallèle.

````
from alibi_detect.cd import TabularDrift

ref = reference[FEATURE_COLUMNS].copy()
cur = analysis[FEATURE_COLUMNS].copy()

categories_per_feature = {}
for col in CATEGORICAL_FEATURES:
    ref[col] = ref[col].astype("category")
    categories = ref[col].cat.categories
    cur[col] = cur[col].astype(pd.CategoricalDtype(categories=categories))
    ref[col] = ref[col].cat.codes
    cur[col] = cur[col].cat.codes
    categories_per_feature[FEATURE_COLUMNS.index(col)] = len(categories)

x_ref = ref.to_numpy(dtype=np.float64)
x_cur = cur.to_numpy(dtype=np.float64)

alibi_detector = TabularDrift(
    x_ref=x_ref,
    p_val=0.05,
    categories_per_feature=categories_per_feature,
    correction="fdr",
)
alibi_prediction = alibi_detector.predict(x_cur, drift_type="feature")

rows = []
for i, col in enumerate(FEATURE_COLUMNS):
    rows.append({
        "feature": col,
        "type": "categorical" if col in CATEGORICAL_FEATURES else "numerical",
        "is_drift": bool(alibi_prediction["data"]["is_drift"][i]),
        "p_value": float(alibi_prediction["data"]["p_val"][i]),
    })

alibi_df = pd.DataFrame(rows)
alibi_df
````

````
dataset_drift = bool(alibi_prediction["data"]["is_drift"].any())
print(f"Dataset drift (au moins une variable en drift) : {dataset_drift}")
````

## 3. NannyML — drift univarié

Catalogue proche d'Alibi-Detect (KS, Chi-deux), mais NannyML gère aussi le découpage temporel (chunking) nativement.

````
import nannyml as nml

univariate_calc = nml.UnivariateDriftCalculator(
    column_names=FEATURE_COLUMNS,
    continuous_methods=["kolmogorov_smirnov"],
    categorical_methods=["chi2"],
    timestamp_column_name=TIMESTAMP_COL,
)
univariate_calc.fit(reference)
drift_results = univariate_calc.calculate(analysis)
drift_df = drift_results.to_df()
drift_df.head()
````

## 4. NannyML — CBPE : estimation de performance sans les vrais labels

Le seul des trois outils qui estime la performance du modèle (ROC-AUC, precision, recall, F1) en production **sans** avoir besoin des vrais labels de fraude, à condition que les probabilités prédites soient bien calibrées. Ne fonctionne bien que sous data drift, pas sous concept drift.

````
estimator = nml.CBPE(
    y_pred_proba=PREDICTED_PROBA_COL,
    y_pred=PREDICTION_COL,
    y_true=TARGET_COL,
    timestamp_column_name=TIMESTAMP_COL,
    metrics=["roc_auc", "precision", "recall", "f1"],
    chunk_period="M",
)
estimator.fit(reference)
performance_results = estimator.estimate(analysis)
performance_df = performance_results.to_df()
performance_df.head()
````

````
performance_results.plot()
````

## Bilan

Compare ici, à la main, ce que chaque outil a détecté :
- Quelles variables Alibi-Detect et NannyML signalent-ils en drift, et sont-elles les mêmes ?
- Le rapport Evidently (`evidently_data_drift.html`) confirme-t-il ces mêmes variables ?
- Le graphique CBPE ci-dessus montre-t-il une baisse de performance estimée qui coïncide avec les variables en drift ?

C'est cette mise en relation (drift observé ↔ dégradation de performance) qui est le cœur du mémoire.
